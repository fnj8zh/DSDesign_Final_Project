# Who Produced the Data and How



The dataset was created and is regularly updated by Andrew Sundberg, a Kaggle contributor who compiled NCAA Division I men’s basketball statistics across twelve seasons (2013–2024) into a single standardized dataset. Sundberg collected team performance data from publicly available NCAA statistical spreadsheets, cleaned and standardized each season’s file, and then merged them into one CSV (cbb.csv). Once complete, he added a year column to show the season associated with each team entry.

This dataset includes a wide range of features, including not just basic statistics such as games played and games won, but also more advanced attributes such as team metadata, efficiency metrics, offensive and defensive performance, shooting percentages, postseason outcomes, and other analytics that are beneficial for tournament analysis, machine learning models, and more.
