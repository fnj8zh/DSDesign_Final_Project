# The Anatomy of a Cinderella Team: What Makes an Upset Possible?

![Description](The%20anatomy%20of%20a%20cinderella%20Team.png)


## Description
This infographic presents a breakdown of what differentiates true NCAA March Madness “Cinderella” teams, which are low-seeded teams (11–16) that reach the Sweet 16, from other low seeds that lose early. Using data from 307 teams across the 2013–2024 tournaments, this graphic takes on three different stages of my analysis: first defining the Cinderella phenomenon in college basketball playoffs, then identifying the strongest differentiating statistics, and finally creating a metric to summarize upset potential and predict future bracket busters. 

The first section introduces the idea of a Cinderella team and the data used. Then the graphic goes directly into comparing low-seeded teams, with the focus on how do teams that advanced further than expected differ from those that didn't statistically. By identifying which variables were of greatest significance, plus TOR for comparison of an insignificant feature, I created a heatmap. This heatmap took normalized statistics grouped by Cinderella or non-Cinderella teams and visualizes z-score differences in performance metrics. The stronger color, the more separation the statistic has between Cinderella teams and other low seeds. This heatmap lays the foundation for the calculation of the “Cinderella Grade,” a metric I made by weighting the most predictive statistics according to their effect sizes and scaling it intuitively (from 1-100).

The middle section visualizes how the Cinderella grade is built using a component breakdown pie chart and explains how the score is normalized to a 0–100 scale. This section allows the reader to understand what is going on in the final plot. 

Finally, the last section features a split violin plot comparing the distribution of Cinderella Grades between Cinderella teams and other low-seeded teams. By its shape and summary statistic dotted lines, this violin plot reveals a clear separation in medians, quartiles, and score spread. It is important to note that the bottom 25% cutoff for past Cinderella team Cinderella Grades is about 70, just below the top 75% cutoff for non-Cinderella teams. Therefore, this metric is a good predictor of what teams have the best chance to become bracket-busters. 

All together, this infographic is a sort of chronological display of my thought process while parsing through this data. It was fascinating to play around with combinations of different statistics and trying different stat weights to optimize the Cinderella Grade. In the future, I see more factors being factored into this grade for even more optimization, as well as more data included.
